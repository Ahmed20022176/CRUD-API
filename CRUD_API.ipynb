```python
%%writefile main.py
from fastapi import FastAPI
from pydantic import BaseModel

# Initialize the FastAPI engine
app = FastAPI()

# Simulate a database using a Python list
todo_list = []

# NOTE: class name must match exactly (capital T) everywhere it's used below
class Task(BaseModel):
    id: int
    title: str
    is_done: bool = False

# Endpoint to retrieve all saved tasks
@app.get("/tasks")
def get_all_tasks():
    return {"tasks": todo_list}

# Endpoint to submit and save a new task
@app.post("/tasks")
def create_task(new_task: Task):
    todo_list.append(new_task)
    return {"message": "Task added successfully!"}

# Endpoint to modify an existing task
@app.put("/tasks/{task_id}")
def update_task(task_id: int, updated_task: Task):
    for i in range(len(todo_list)):
        if todo_list[i].id == task_id:
            todo_list[i] = updated_task
            return {"message": "Task updated!"}
    return {"error": "Task not found"}

# Endpoint to permanently delete a task
@app.delete("/tasks/{task_id}")
def delete_task(task_id: int):
    for i in range(len(todo_list)):
        if todo_list[i].id == task_id:
            del todo_list[i]
            return {"message": "Task deleted!"}
    return {"error": "Task not found"}
```
